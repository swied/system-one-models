```python
import os, sys, time
from pathlib import Path
from dotenv import load_dotenv
from typesafe_sdk import Choice, Noul, Score, TypeSafeClient
```

```python
# Load environment variables from a .env file
env_file = Path.home() / ".config" / "typesafe" / "secrets.env"
load_dotenv(env_file)

# Get the API key from the environment variable
api_key = os.environ["TYPESAFE_API_KEY"]

# Initialize the TypeSafe client with the API key
client = TypeSafeClient()
```

```python
# 1. Record the start time
start_time = time.perf_counter()

ticket = "Hi, I've been trying to connect my Stripe account for 3 days and the integration keeps failing. I'm losing sales. Please help ASAP."

response = client.system_one(
    state=ticket,
    questions={
        "department": Choice(
            instructions="Which team should handle this",
            criteria={
                "billing": "Payment or subscription issues",
                "technical": "Bugs or integration problems",
                "sales": "Pricing or account questions",
            },
        ),
        "frustration": Score(
            instructions="How frustrated the customer appears",
            criteria=[
                "Calm, just stating facts",
                "Frustrated but civil",
                "Very angry, strong language",
            ],
        ),
        "is_urgent": Noul(
            instructions="The message conveys urgency or time-sensitivity",
        ),
    },
)

end_time = time.perf_counter()

# 4. Calculate and format the elapsed time
execution_time = end_time - start_time

print(f"Execution time: {execution_time:.6f} seconds")
print("department:", response.answers["department"].choice)  # "technical"
print("frustration:", response.answers["frustration"].score)  # 1.0
print("is_urgent:", response.answers["is_urgent"].noul)     # 1.0
```
