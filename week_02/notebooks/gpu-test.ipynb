```python
import torch

print(f"PyTorch Version: {torch.__version__}")
```

```python
# 1. Check if CUDA (NVIDIA GPU support) or MPS (Apple Silicon support) is available
cuda_available = torch.cuda.is_available()
mps_available = hasattr(torch.backends, "mps") and torch.backends.mps.is_available()

if cuda_available:
    device_type = "cuda"
    print("\n🟢 Success: NVIDIA GPU detected via CUDA!")
    print(f"Number of GPUs: {torch.cuda.device_count()}")
    print(f"Active GPU Name: {torch.cuda.get_device_name(0)}")
elif mps_available:
    device_type = "mps"
    print("\n🟢 Success: Apple Silicon GPU detected via MPS!")
else:
    device_type = "cpu"
    print("\n🔴 Failed: No GPU detected. PyTorch is defaulting to CPU.")
    print("If you have a dedicated GPU, you may need to reinstall the correct PyTorch package.")
```

```python
# 2. Perform a basic mathematical operation to ensure the GPU is functional
if device_type != "cpu":
    try:
        # Define the target device
        device = torch.device(device_type)
        
        # Create two random matrices directly on the GPU
        x = torch.rand(3, 3, device=device)
        y = torch.rand(3, 3, device=device)
        
        # Perform matrix multiplication on the GPU
        result = torch.matmul(x, y)
        
        print(f"\n⚡ Calculation test passed!")
        print(f"Tensor is successfully residing on: {result.device}")
    except Exception as e:
        print(f"\n❌ Error during GPU computation: {e}")
```
